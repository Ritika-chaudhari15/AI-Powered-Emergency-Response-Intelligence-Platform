# Feature Engineering

## Objective

Create engineered features for model training.

Features created:

- WeatherScore
- InfrastructureScore

Output:

final_disaster_dataset_engineered.csv

```python
from pathlib import Path

import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler
```

```python
PROJECT_DIR = Path.cwd().parent

DATASET_DIR = PROJECT_DIR / "datasets"

INPUT_FILE = (
    DATASET_DIR /
    "final_disaster_dataset_with_infrastructure.csv"
)

OUTPUT_FILE = (
    DATASET_DIR /
    "final_disaster_dataset_engineered.csv"
)
```

```python
df = pd.read_csv(INPUT_FILE)

print(df.shape)

df.head()
```

```python
weather_scaler = MinMaxScaler()

weather_columns = [

    "Temperature",

    "Humidity",

    "Rainfall"

]

df[weather_columns] = weather_scaler.fit_transform(
    df[weather_columns]
)
```

```python
df["WeatherScore"] = (

    df["Temperature"]

    + df["Humidity"]

    + df["Rainfall"]

) / 3
```

```python
distance_columns = [

    "NearestHospitalDistanceKm",

    "NearestClinicDistanceKm",

    "NearestSchoolDistanceKm",

    "NearestFireStationDistanceKm"

]

count_columns = [

    "HospitalCount25Km",

    "ClinicCount25Km",

    "SchoolCount25Km",

    "FireStationCount25Km"

]
```

```python
distance_scaler = MinMaxScaler()

df[distance_columns] = distance_scaler.fit_transform(
    df[distance_columns]
)

# Smaller distance is better,
# so invert the values.

df[distance_columns] = 1 - df[distance_columns]
```

```python
count_scaler = MinMaxScaler()

df[count_columns] = count_scaler.fit_transform(
    df[count_columns]
)
```

```python
df["InfrastructureScore"] = (

    df[distance_columns].mean(axis=1)

    +

    df[count_columns].mean(axis=1)

) / 2
```

```python
print(

    df[

        [

            "WeatherScore",

            "InfrastructureScore"

        ]

    ].describe()

)
```

```python
df.to_csv(

    OUTPUT_FILE,

    index=False

)

print()

print("Saved Successfully")

print()

print(OUTPUT_FILE)
```
