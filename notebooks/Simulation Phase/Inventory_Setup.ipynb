```python
# ============================================================
# Resources Available At Every Warehouse
# ============================================================

resource_template = {

    "Food Packets": 2_000_000,

    "Water Bottles": 3_000_000,

    "Medical Kits": 400_000,

    "Blankets": 600_000,

    "Temporary Tents": 80_000,

    "Rescue Teams": 300,

    "Ambulances": 300,

    "Trucks": 300,

    "Boats": 80,

    "Helicopters": 25

}
```

```python
inventory_rows = []

for warehouse in warehouses:

    for resource, quantity in resource_template.items():

        inventory_rows.append({

            "WarehouseID": warehouse["WarehouseID"],

            "WarehouseName": warehouse["WarehouseName"],

            "State": warehouse["State"],

            "Latitude": warehouse["Latitude"],

            "Longitude": warehouse["Longitude"],

            "Resource": resource,

            "AvailableQuantity": quantity

        })

inventory = pd.DataFrame(inventory_rows)

inventory.head()
```
