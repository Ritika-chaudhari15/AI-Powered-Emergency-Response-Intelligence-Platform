# 03 - Merge Infrastructure Features

## Objective

Merge the disaster dataset with the infrastructure features generated in
Notebook 2.

Input Files

- final_disaster_dataset.csv
- event_infrastructure_summary.csv

Output

- final_disaster_dataset_with_infrastructure.csv

This dataset will be used for:

- Regression Model Training
- Resource Allocation
- Relief Coordination

```python
from pathlib import Path

import pandas as pd

import numpy as np
```

```python
PROJECT_DIR = Path.cwd().parent

DATASET_DIR = PROJECT_DIR / "datasets"

DISASTER_DATASET = (
    DATASET_DIR /
    "final_disaster_dataset.csv"
)

INFRASTRUCTURE_DATASET = (
    DATASET_DIR /
    "event_infrastructure_summary.csv"
)

OUTPUT_DATASET = (
    DATASET_DIR /
    "final_disaster_dataset_with_infrastructure.csv"
)
```

```python
disaster_df = pd.read_csv(
    DISASTER_DATASET
)

infra_df = pd.read_csv(
    INFRASTRUCTURE_DATASET
)

print()

print("Disaster Dataset")

print(disaster_df.shape)

print()

print("Infrastructure Dataset")

print(infra_df.shape)
```

```python
print()

print(
    "Unique Disaster IDs:",
    disaster_df["SourceEventID"].nunique()
)

print()

print(
    "Unique Infrastructure IDs:",
    infra_df["SourceEventID"].nunique()
)
```

```python
missing = set(
    disaster_df["SourceEventID"]
) - set(
    infra_df["SourceEventID"]
)

print()

print("Missing IDs:")

print(len(missing))
```

```python
final_df = disaster_df.merge(

    infra_df,

    on="SourceEventID",

    how="left"

)

print(final_df.shape)
```

```python
print()

print(
    final_df.isna().sum()
)
```

```python
print()

print(final_df.head())

print()

print(final_df.info())
```

```python
columns = [

    "NearestHospitalDistanceKm",

    "HospitalCount25Km",

    "NearestClinicDistanceKm",

    "ClinicCount25Km",

    "NearestSchoolDistanceKm",

    "SchoolCount25Km",

    "NearestFireStationDistanceKm",

    "FireStationCount25Km"

]

print()

print(

    final_df[
        columns
    ].describe()

)
```

```python
final_df.to_csv(

    OUTPUT_DATASET,

    index=False

)

print()

print("Saved Successfully")

print()

print(OUTPUT_DATASET)
```
