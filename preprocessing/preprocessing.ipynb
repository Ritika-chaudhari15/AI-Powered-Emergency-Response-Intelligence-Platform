```python
df.drop_duplicates(inplace=True)
df = df[df["Disaster Type"] == "Earthquake"]

df = df.dropna(
    subset=[
        "Magnitude",
        "Latitude",
        "Longitude"
    ]
)
```

```python
df["Total Affected"] = df["Total Affected"].fillna(
    df["Total Affected"].median()
)

# Fill missing Location
df["Location"] = df["Location"].fillna("Unknown")
```
